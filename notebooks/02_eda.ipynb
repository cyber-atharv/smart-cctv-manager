# Notebook 02: Exploratory Data Analysis (EDA)
## Smart CCTV AI Safety & Identification Platform

In this notebook, we analyze:
- Feature distributions across postures (Standing, Sitting, Working, Falling)
- Aspect ratio and body inclination angle distributions
- Kinematic vertical velocity during falls
- Feature correlation matrices

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = '../data/raw/fall_activity_dataset.csv'
df = pd.read_csv(DATA_PATH)
sns.set_theme(style="whitegrid")
```

```python
# 1. Body Angle & Aspect Ratio Distribution across Activities
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(x='label', y='body_angle', data=df, ax=axes[0], palette='Set2')
axes[0].set_title('Body Torso Angle (deg) by Activity Class')
axes[0].set_ylabel('Body Angle (degrees from vertical)')

sns.boxplot(x='label', y='height_width_ratio', data=df, ax=axes[1], palette='Set2')
axes[1].set_title('Bounding Box Aspect Ratio (Height/Width) by Activity Class')
axes[1].set_ylabel('Height / Width Ratio')

plt.tight_layout()
plt.show()
```

```python
# 2. Vertical Velocity vs Movement Speed Scatter
plt.figure(figsize=(9, 6))
sns.scatterplot(x='movement_speed', y='vertical_velocity', hue='label', data=df, alpha=0.7, palette='tab10')
plt.title('Kinematic Profile: Movement Speed vs Vertical Velocity')
plt.xlabel('Movement Speed (norm/frame)')
plt.ylabel('Vertical Velocity (downward +)')
plt.show()
```

```python
# 3. Feature Correlation Matrix Heatmap
numeric_cols = df.select_dtypes(include=[np.number]).columns
plt.figure(figsize=(12, 9))
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=False, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Feature Pearson Correlation Matrix')
plt.show()
```
